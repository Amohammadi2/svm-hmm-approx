```python
import pandas as pd
import numpy as np
from dataloader import load_gold18_history_data

df = load_gold18_history_data()

df.head(20)
```

```python
df.describe()
```

```python
df["Log_Return"] = np.log(df["Final"] / df["Final"].shift(1)) * 100
df.head()
```

```python

df = df.sort_index(ascending=True)
df.reset_index(drop=True, inplace=True)
```

```python
df.head()
```

```python
#visualize the returns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.plot(df.index, df['Log_Return'])
plt.xlabel('Time')
plt.ylabel('Log Returns')
plt.title('Gold Price Log Returns')
plt.show()
```

```python
df.to_csv('./gold18_history_log_returns.csv', columns=['Log_Return'], index=True)
```
